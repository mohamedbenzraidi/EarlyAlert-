```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/chartevents.csv")
patient = df[df["subject_id"] == 1]

plt.plot(patient["charttime"], patient["itemid"]=="HR")
plt.plot(patient["charttime"], patient["itemid"]=="SpO2")
plt.legend(["HR", "SpO2"])
plt.show()
```
